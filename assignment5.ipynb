```python
#Q2
import numpy as np
#a)
arr = np.array([1,2,3,6,4,5])

rev = arr[::-1]
print("Original array:", arr)
print("Reversed array: ",rev)

#b)
x= np.array([1,2,3,4,5,1,2,1,1,1])
values,counts = np.unique(x,return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x==most_frequent)[0]
print("Most frequent value:" ,most_frequent)
print("indices: ", indices)

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values,counts = np.unique(y,return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y==most_frequent)[0]
print("Most frequent value:" ,most_frequent)
print("indices: ", indices)
```

```python
#Q4
import numpy as np

Aditya = np.linspace(10,100,25)
print("array:")
print(Aditya)
print("Dimensions: ", Aditya.ndim)
print("Shape: ", Aditya.shape)
print("Total Elements: ", Aditya.size)
print("Data type: ", Aditya.dtype)
print("Total bytes: ", Aditya.nbytes)

transpose = Aditya.reshape(1,-1)
print("transpose: ")
print(transpose)

print("transpose using T:")
print(Aditya.T)
```

```python
#Q5
import numpy as np
ucs420_Aditya = np.array([[10,20,30,40],[50,60,70,80],[90,15,20,35]])

print("original array:")
print(ucs420_Aditya)

print("Mean: ", np.mean(ucs420_Aditya))

print("Median:", np.median(ucs420_Aditya))

print("Maximum:", np.max(ucs420_Aditya))

print("Minimum:", np.min(ucs420_Aditya))

print("Unique elements:", np.unique(ucs420_Aditya))

reshaped_ucs420_Aditya = ucs420_Aditya.reshape(4, 3)

print("Reshaped Array (4 x 3):")
print(reshaped_ucs420_Aditya)

# Resize into 2 rows and 3 columns
resized_ucs420_Aditya = np.resize(ucs420_Aditya, (2, 3))

print("Resized Array (2 x 3):")
print(resized_ucs420_Aditya)
```
