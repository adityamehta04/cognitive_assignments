```python
#Q6
import pandas as pd

data = {
    "Employee_ID":[101,102,103,104,105],
    "Name":["Alice","Bob","Charlie","Diana","Edward"],
    "Department":["HR","IT","IT","Marketing","Sales"],
    "Age":[29,34,41,28,38],
    "Salary":[50000,70000,65000,55000,60000],
    "Years_of_Experience":[4,8,10,3,12],
    "Joining_Date":["2020-03-15","2017-07-19","2013-06-01","2021-02-10","2010-11-25"],
    "Gender":["Female","Male","Male","Female","Male"],
    "Bonus":[5000,7000,6000,4500,5000],
    "Rating":[4.5,4.0,3.8,4.7,3.5]
}

df = pd.DataFrame(data)
df.to_csv("employees.csv", index=False)

print(df)
```

```python
#a
import pandas as pd

df = pd.read_csv("employees.csv")

print(df.shape)

#b
print(df.info())

#c
print(df.describe())

#d
print("First 5 rows")
print(df.head())

print("\nLast 3 rows")
print(df.tail(3))

#e
#1.
print("Average Salary =", df["Salary"].mean())

#2.
print("Total Bonus =", df["Bonus"].sum())

#3.
print("Youngest Age =", df["Age"].min())

#4.
print("Highest Rating =", df["Rating"].max())

#f
sorted_df = df.sort_values(by="Salary", ascending=False)

print(sorted_df)

#g
def category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

df["Performance"] = df["Rating"].apply(category)

print(df)

#h
print(df.isnull())

print("\nMissing values in each column")
print(df.isnull().sum())

#i
df.rename(columns={"Employee_ID":"ID"}, inplace=True)

print(df.head())

#j
#1
print(df[df["Years_of_Experience"] > 5])
#2
print(df[df["Department"] == "IT"])

#k
df["Tax"] = df["Salary"] * 0.10

print(df)

#l
df.to_csv("employees_modified.csv", index=False)

print("Modified dataset saved successfully.")
```
