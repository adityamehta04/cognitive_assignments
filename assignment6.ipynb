```python
#Q1
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

#a)
corrected_temperature = temperature + 2
print("a) Corrected temperatures:", corrected_temperature)

#b)
fahrenheit = (9 / 5) * temperature + 32
print("b) Temperatures in Fahrenheit:", fahrenheit)

#c)
readings_greater_32 = temperature[temperature > 32]
print("c) Readings greater than 32°C:", readings_greater_32)

#d)
count = np.sum(temperature > 32)
print("d) Count:", count)

#e)
print("e)")
print("Vectorization : Performs operations on the entire NumPy array at once, avoiding explicit for loops and generally improving performance.")
print("Boolean indexing: Uses a condition such as temperature > 32 to select only the required elements, making filtering concise and efficient.")
```

```python
#Q2
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

#a)
print("a) Total steps:", np.sum(steps))

#b)
print("b) Mean steps:", np.mean(steps))

#c)
print("c)")
print("Maximum:", np.max(steps))
print("Minimum:", np.min(steps))

#d)
print("d) Total per day:", np.sum(steps, axis=0))

#e)
print("e) Total per user:", np.sum(steps, axis=1))

#f)
position = np.unravel_index(np.argmax(steps), steps.shape)
print("f) Maximum position:", position)
```

```python
#Q3
import numpy as np

#a)
original = np.array([1, 2, 3, 4, 5, 6])

#b)
subset = original[1:4]

#c)
subset[0] = 999

print("Original after modifying subset:", original)
print("Subset:", subset)

#d)
copied = original[1:4].copy()
copied[0] = 500
print("d)")
print("Original after modifying copied:", original)
print("Copied array:", copied)

#e)
matrix = np.arange(1, 13).reshape(3, 4)
print("e)")
print("3 x 4 matrix:\n", matrix)

#f)
print("f)")
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:\n", matrix[1:3, 2:4])

#g)
flat_array = matrix.flatten()
ravel_array = matrix.ravel()
print("g)")
print("Flatten:", flat_array)
print("Ravel:", ravel_array)

#h)
ravel_array[0] = 100
print("h)")
print("Matrix after modifying ravel:", matrix)

#i)
flat_array[1] = 200
print("i)")
print("Matrix after modifying flatten:", matrix)

#j)
print("j)")
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
#Q4
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
], dtype=float)

y = np.array([40, 65, 30, 85], dtype=float)

#a)
print("a)")
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

#b)
print("b)")
print("Transpose of X:\n", X.T)

#c)
XTX = X.T @ X
print("c)")
print("X.T @ X:\n", XTX)

#d)
XTX_inv = np.linalg.inv(XTX)
print("d)")
print("Inverse of X.T @ X:\n", XTX_inv)

#e)
beta = XTX_inv @ X.T @ y
print("e)")
print("Regression coefficients:", beta)

#g)
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("g)")
print("Predicted assistance score:", predicted_score)
```
